```python
inpatient_claims = (
    inpatient.groupby("Provider")["ClaimID"]
    .nunique()
    .rename("InpatientClaims")
)

outpatient_claims = (
    outpatient.groupby("Provider")["ClaimID"]
    .nunique()
    .rename("OutpatientClaims")
)

claim_features = pd.concat(
    [inpatient_claims, outpatient_claims],
    axis=1
).fillna(0)

claim_features["TotalClaims"] = (
    claim_features["InpatientClaims"]
    + claim_features["OutpatientClaims"]
)
```

```python
eda_df = provider.merge(
    claim_features,
    left_on="Provider",
    right_index=True,
    how="left"
)
```

```python
diagnosis_cols_in = [
    col for col in inpatient.columns
    if col.startswith("ClmDiagnosisCode")
]

diagnosis_cols_out = [
    col for col in outpatient.columns
    if col.startswith("ClmDiagnosisCode")
]

print("Inpatient diagnosis columns:", diagnosis_cols_in)
print("Outpatient diagnosis columns:", diagnosis_cols_out)
```

```python
inpatient_diagnosis_count = (
    inpatient[diagnosis_cols_in]
    .notna()
    .sum(axis=1)
    .sum()
)

outpatient_diagnosis_count = (
    outpatient[diagnosis_cols_out]
    .notna()
    .sum(axis=1)
    .sum()
)

print("Inpatient diagnosis entries:", inpatient_diagnosis_count)
print("Outpatient diagnosis entries:", outpatient_diagnosis_count)
```

```python

inpatient["DiagnosisEntries"] = (
    inpatient[diagnosis_cols_in].notna().sum(axis=1)
)

outpatient["DiagnosisEntries"] = (
    outpatient[diagnosis_cols_out].notna().sum(axis=1)
)

inpatient_diag = inpatient.groupby("Provider")["DiagnosisEntries"].sum()
outpatient_diag = outpatient.groupby("Provider")["DiagnosisEntries"].sum()

total_diagnosis_entries = (
    inpatient_diag.add(outpatient_diag, fill_value=0)
)


model_df["TotalDiagnosisEntries"] = (
    model_df["Provider"].map(total_diagnosis_entries).fillna(0)
)

model_df["DiagnosisIntensity"] = (
    model_df["TotalDiagnosisEntries"] / model_df["TotalClaims"]
)

model_df[[
    "Provider",
    "TotalClaims",
    "TotalDiagnosisEntries",
    "DiagnosisIntensity"
]].head()
```

```python
procedure_cols_in = [
    col for col in inpatient.columns
    if col.startswith("ClmProcedureCode")
]

procedure_cols_out = [
    col for col in outpatient.columns
    if col.startswith("ClmProcedureCode")
]

print("Inpatient procedure columns:", procedure_cols_in)
print("Outpatient procedure columns:", procedure_cols_out)
```

```python
inpatient["ProcedureEntries"] = (
    inpatient[procedure_cols_in].notna().sum(axis=1)
)

outpatient["ProcedureEntries"] = (
    outpatient[procedure_cols_out].notna().sum(axis=1)
)

print("Inpatient procedure entries:",
      inpatient["ProcedureEntries"].sum())

print("Outpatient procedure entries:",
      outpatient["ProcedureEntries"].sum())
```

```python

inpatient_proc = (
    inpatient.groupby("Provider")["ProcedureEntries"].sum()
)

outpatient_proc = (
    outpatient.groupby("Provider")["ProcedureEntries"].sum()
)


total_procedure_entries = (
    inpatient_proc.add(outpatient_proc, fill_value=0)
)


model_df["TotalProcedureEntries"] = (
    model_df["Provider"].map(total_procedure_entries).fillna(0)
)


model_df["ProcedureIntensity"] = (
    model_df["TotalProcedureEntries"] / model_df["TotalClaims"]
)

model_df[[
    "Provider",
    "TotalClaims",
    "TotalProcedureEntries",
    "ProcedureIntensity"
]].head()
```

```python
model_df["ClaimsPerPhysician"] = (
    model_df["TotalClaims"]
    / model_df["UniquePhysicianCount"].replace(0, np.nan)
)

model_df[[
    "Provider",
    "TotalClaims",
    "UniquePhysicianCount",
    "ClaimsPerPhysician"
]].head()
```

```python
target = "PotentialFraud"

exclude_cols = ["Provider", target]

X = model_df.drop(columns=exclude_cols)
y = model_df[target].map({"No": 0, "Yes": 1})

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
print("\nTarget proportions:")
print(y.value_counts(normalize=True).round(3))
```

```python
from sklearn.model_selection import train_test_split

X_train_full, X_test_raw, y_train_full, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train = X_train_full.copy()
X_test = X_test_raw.copy()
y_train = y_train_full.copy()

print("Train:", X_train_full.shape)
print("Test:", X_test_raw.shape)
```

```python
from sklearn.impute import SimpleImputer

missing_features = [
    "AvgLengthOfStay",
    "MedianLengthOfStay",
    "UniqueProcedureCount"
]

for col in missing_features:
    X_train[col + "_Missing"] = X_train[col].isna().astype(int)
    X_test[col + "_Missing"] = X_test[col].isna().astype(int)

imputer = SimpleImputer(strategy="median")

X_train[missing_features] = imputer.fit_transform(
    X_train[missing_features]
)

X_test[missing_features] = imputer.transform(
    X_test[missing_features]
)

print("Remaining missing values in train:", X_train.isna().sum().sum())
print("Remaining missing values in test:", X_test.isna().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    class_weight="balanced",
    max_iter=2000,
    random_state=42
)

lr_model.fit(X_train_scaled, y_train)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

y_pred = lr_model.predict(X_test_scaled)
y_proba = lr_model.predict_proba(X_test_scaled)[:, 1]

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("ROC-AUC:", round(roc_auc_score(y_test, y_proba), 4))
print("PR-AUC (Average Precision):",round(average_precision_score(y_test, y_proba), 4))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

rf_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_proba = rf_model.predict_proba(X_test)[:, 1]

print("Confusion Matrix:")
print(confusion_matrix(y_test, rf_pred))

print("\nClassification Report:")
print(classification_report(y_test, rf_pred, zero_division=0))

print("ROC-AUC:", round(roc_auc_score(y_test, rf_proba), 4))
print("PR-AUC:", round(average_precision_score(y_test, rf_proba), 4))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_val,
    xgb_val_proba
)


precision_at_threshold = precision[:-1]
recall_at_threshold = recall[:-1]

results = pd.DataFrame({
    "threshold": thresholds,
    "precision": precision_at_threshold,
    "recall": recall_at_threshold
})

candidates = results[results["recall"] >= 0.80]

candidates = candidates.sort_values(
    "precision",
    ascending=False
)

print(candidates.head(10))
```

```python
if candidates.empty:
    print("No threshold achieved 80% recall.")
else:
    best_row = candidates.iloc[0]
    best_threshold = best_row["threshold"]

    print("Selected threshold:", best_threshold)
    print("Validation precision:", best_row["precision"])
    print("Validation recall:", best_row["recall"])
```
